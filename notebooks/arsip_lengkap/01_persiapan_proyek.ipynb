## 1. Pemeriksaan Environment & Hardware Accelerator (GPU)

```python
import sys
import platform
import numpy as np
import pandas as pd
import torch
import torchvision

print("=" * 60)
print("🔍 PEMERIKSAAN ENVIRONMENT - PROYEK CITRA SERUPA")
print("=" * 60)

# 1. Versi Python & Sistem Operasi
print(f"🐍 Versi Python      : {sys.version.split()[0]}")
print(f"💻 Sistem Operasi    : {platform.system()} {platform.release()} ({platform.machine()})")

# 2. Pustaka Dasar
print(f"📦 Versi NumPy       : {np.__version__}")
print(f"📦 Versi Pandas      : {pd.__version__}")
print(f"🔥 Versi PyTorch     : {torch.__version__}")
print(f"👁️  Versi TorchVision : {torchvision.__version__}")

# 3. Ketersediaan Akselerator GPU
print("-" * 60)
gpu_tersedia = torch.cuda.is_available()
print(f"⚡ GPU Tersedia      : {'YA ✅' if gpu_tersedia else 'TIDAK ❌ (Menggunakan CPU)'}")

if gpu_tersedia:
    print(f"🎮 Nama GPU          : {torch.cuda.get_device_name(0)}")
    print(f"🧠 Memori GPU        : {torch.cuda.get_device_properties(0).total_memory / (1024**3):.2f} GB")
else:
    print("💡 Catatan: Untuk tugas visual similarity deep learning, aktifkan GPU via:")
    print("   Menu Runtime -> Change runtime type -> Pilih T4 GPU -> Save")
print("=" * 60)
```

## 2. Menghubungkan Google Colab ke Google Drive

```python
from google.colab import drive
import os

# Mount Google Drive ke direktori /content/drive
drive.mount('/content/drive')

# Cek lokasi folder proyek CITRA_SERUPA
drive_path = '/content/drive/MyDrive/CITRA_SERUPA'
alt_path = '/content/drive/MyDrive/CITRA-SERUPA'

target_path = None
if os.path.exists(drive_path):
    target_path = drive_path
elif os.path.exists(alt_path):
    target_path = alt_path

if target_path:
    print(f"\n✅ Berhasil terhubung ke Google Drive!")
    print(f"📁 Folder Proyek : {target_path}")
    print("📂 Subfolder     :", os.listdir(target_path))
else:
    print("\n⚠️ Folder CITRA_SERUPA tidak ditemukan di root MyDrive.")
    print("Pastikan folder CITRA_SERUPA ada di Google Drive akun yang terhubung.")
```
